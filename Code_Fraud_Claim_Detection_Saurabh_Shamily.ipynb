## **2. Data Cleaning** <font color = red>[10 marks]</font>

### **2.1 Handle null values** <font color = red>[2 marks]</font>

#### **2.1.1** Examine the columns to determine if any value or column needs to be treated <font color="red">[1 Mark]</font>

````python
# Check the number of missing values in each column

# First check for NULL/NaN values
null_counts = df.isnull().sum()
print("Number of NULL values in each column:")
print(null_counts[null_counts > 0])

# Then check for placeholder missing values ('?')
print("\nChecking for '?' placeholders:")
placeholder_counts = {}
for column in df.select_dtypes(include=['object']).columns:
    question_marks = (df[column] == '?').sum()
    if question_marks > 0:
        placeholder_counts[column] = question_marks

if placeholder_counts:
    for col, count in placeholder_counts.items():
        print(f"{col}: {count}")
else:
    print("No '?' placeholders found.")

# Display columns with any type of missing value
print("\nTotal missing values (NULL or '?') in each column:")
total_missing = null_counts.copy()
for col, count in placeholder_counts.items():
    total_missing[col] = total_missing[col] + count
print(total_missing[total_missing > 0])
````

The above output summarizes the identification of missing values in the dataset. It indicates that missing values are present in two forms: standard NULL/NaN values and the placeholder string `'?'` in object columns.

The columns identified with missing values and their respective total counts (combining NULL/NaN and '?') are:

*   `_c39`: 1000 missing values (all NULL/NaN).
*   `property_damage`: 360 missing values (all '?').
*   `police_report_available`: 343 missing values (all '?').
*   `collision_type`: 178 missing values (all '?').
*   `authorities_contacted`: 91 missing values (all NULL/NaN).

The `_c39` column is entirely missing values, while the other listed columns have partial missingness that requires handling.

#### **2.1.2** Handle rows containing null values <font color="red">[1 Mark]</font>

````python
# Handle the rows containing null values

# First, drop the completely empty column _c39
df = df.drop('_c39', axis=1)

# Replace '?' placeholders with NaN
for column in df.select_dtypes(include=['object']).columns:
    df[column] = df[column].replace('?', np.nan)

# Check percentage of missing values in each column
missing_percentage = (df.isnull().sum() / len(df)) * 100
print("Percentage of missing values in each column after replacing placeholders:")
print(missing_percentage[missing_percentage > 0])

# Strategy for authorities_contacted: impute with most frequent value (mode) since this is a categorical variable
if df['authorities_contacted'].isnull().sum() > 0:
    most_common = df['authorities_contacted'].mode()[0]
    df['authorities_contacted'] = df['authorities_contacted'].fillna(most_common)
    print(f"\nImputed missing values in 'authorities_contacted' with most common value: {most_common}")

# For other columns with placeholders that are now NaN
# Inspect columns with missing values to decide on appropriate treatment
for column in df.columns:
    if df[column].isnull().sum() > 0:
        if df[column].dtype == 'object':  # For categorical variables
            most_common = df[column].mode()[0]
            df[column] = df[column].fillna(most_common)
            print(f"Imputed missing values in '{column}' with most common value: {most_common}")

# Verify no missing values remain
print("\nRemaining missing values after treatment:")
print(df.isnull().sum()[df.isnull().sum() > 0])
````

The provided output details the steps taken to handle missing values in the dataset.

1.  **Column Dropping:** The column `_c39`, identified in the previous step as having 1000 missing values (representing all rows), was dropped from the DataFrame.
2.  **Placeholder Replacement:** The placeholder character `'?'` in object type columns (`collision_type`, `property_damage`, `police_report_available`) was replaced with `np.nan` to standardize the representation of missing values.
3.  **Missing Value Percentage Calculation:** After replacing `'?'`, the percentage of missing values per column was calculated and displayed. This confirmed the presence of `NaN` values in `collision_type` (17.8%), `authorities_contacted` (9.1%), `property_damage` (36.0%), and `police_report_available` (34.3%).
4.  **Imputation Strategy:** Missing values in these columns were handled using imputation. Specifically, for categorical columns (object dtype), the missing values were filled with the mode (most frequent value) of that respective column. The output explicitly states the mode used for imputation for `authorities_contacted` ('Police'), `collision_type` ('Rear Collision'), `property_damage` ('NO'), and `police_report_available` ('NO').
5.  **Verification:** A final check for missing values across all columns confirms that no `NaN` values remain in the DataFrame after the imputation process.

In summary, the `_c39` column was removed due to complete missingness, `'?'` placeholders were standardized to `NaN`, and missing values in relevant categorical columns were imputed using the mode. The dataset now contains no identified missing values.

### **2.2 Identify and handle redundant values and columns** <font color = red>[5 marks]</font>

#### **2.2.1** Examine the columns to determine if any value or column needs to be treated <font color="red">[2 Mark]</font>

````python
# Write code to display all the columns with their unique values and counts and check for redundant values


# Function to analyze columns and print detailed information
def analyze_column(df, column_name):
    """Analyze a column's unique values and distribution"""
    n_unique = df[column_name].nunique()
    value_counts = df[column_name].value_counts()
    
    print(f"\n{'-'*50}")
    print(f"Column: {column_name}")
    print(f"Number of unique values: {n_unique}")
    print(f"Percentage of unique values: {n_unique / len(df) * 100:.2f}%")
    
    # Print the top values and their frequencies
    if n_unique <= 20:  # If fewer than 20 unique values, show all
        print("Value counts:")
        print(value_counts)
    else:  # Otherwise show just the top 10
        print("Top 10 value counts:")
        print(value_counts.head(10))
        
    # Check if column might be an identifier
    if n_unique > len(df) * 0.9:
        print("⚠️ This column appears to be an identifier (high cardinality).")
    
    # Check if column has only one unique value
    elif n_unique == 1:
        print("⚠️ This column has only one unique value and is redundant.")
    
    # Check if column has very low cardinality
    elif n_unique == 2 and df[column_name].dtype != 'bool':
        print("ℹ️ This column has binary values and might be a candidate for encoding.")

# Loop through each column and analyze
print("Analyzing columns for redundancy and unique value distribution:")

# First look at numerical columns
print("\n===== NUMERICAL COLUMNS =====")
for column in df.select_dtypes(include=['number']).columns:
    analyze_column(df, column)

# Then look at categorical columns
print("\n===== CATEGORICAL COLUMNS =====")
for column in df.select_dtypes(include=['object']).columns:
    analyze_column(df, column)

# Summary of potentially redundant or problematic columns
print("\n===== SUMMARY OF POTENTIAL ISSUES =====")
redundant_cols = []
identifier_cols = []
binary_cols = []

for column in df.columns:
    n_unique = df[column].nunique()
    if n_unique == 1:
        redundant_cols.append(column)
    elif n_unique > len(df) * 0.9:
        identifier_cols.append(column)
    elif n_unique == 2 and df[column].dtype != 'bool':
        binary_cols.append(column)

print(f"Redundant columns (1 unique value): {redundant_cols}")
print(f"Potential identifier columns (high cardinality): {identifier_cols}")
print(f"Binary columns (potential for encoding): {binary_cols}")
````

#### **2.2.3** Identify and drop rows where features have illogical or invalid values, such as negative values for features that should only have positive values <font color="red">[1 Mark]</font>

````python
# Identify and drop rows where features have illogical or invalid values, such as negative values for features that should only have positive values

# Record the initial row count
print(f"Initial number of rows: {len(df)}")

# Define columns that should only have positive or zero values
positive_cols = [
    'policy_annual_premium', 
    'total_claim_amount', 
    'injury_claim', 
    'property_claim', 
    'vehicle_claim',
    'bodily_injuries',
    'witnesses'
]

# Check for negative values in columns that should be positive
negative_value_counts = {}
for col in positive_cols:
    negative_count = (df[col] < 0).sum()
    if negative_count > 0:
        negative_value_counts[col] = negative_count

print("\nColumns with negative values that should be positive:")
if negative_value_counts:
    for col, count in negative_value_counts.items():
        print(f"{col}: {count} rows with negative values")
else:
    print("No negative values found in columns that should be positive.")

# Check umbrella_limit specifically as it had a negative value
if (df['umbrella_limit'] < 0).any():
    neg_umbrella_count = (df['umbrella_limit'] < 0).sum()
    print(f"\n'umbrella_limit' column has {neg_umbrella_count} negative value(s)")
    print("Example of negative umbrella_limit row:")
    display(df[df['umbrella_limit'] < 0].head(1))

# Check for other potential illogical values
print("\nChecking for other illogical values:")

# Age should be reasonable for insurance holders (e.g., > 16)
if (df['age'] < 16).any():
    print(f"Found {(df['age'] < 16).sum()} rows with age below 16")

# Auto year shouldn't be in the future
current_year = pd.Timestamp.now().year
if (df['auto_year'] > current_year).any():
    print(f"Found {(df['auto_year'] > current_year).sum()} rows with future auto_year")

# Identify all rows with invalid values to drop
invalid_mask = pd.Series(False, index=df.index)

# Add mask for negative values in positive-only columns
for col in positive_cols:
    invalid_mask = invalid_mask | (df[col] < 0)

# Add mask for negative umbrella_limit
invalid_mask = invalid_mask | (df['umbrella_limit'] < 0)

# Add mask for unreasonable age
invalid_mask = invalid_mask | (df['age'] < 16)

# Add mask for future auto_year
invalid_mask = invalid_mask | (df['auto_year'] > current_year)

# Count total rows to drop
rows_to_drop = invalid_mask.sum()
print(f"\nTotal rows with illogical values to drop: {rows_to_drop}")

# Drop the invalid rows
if rows_to_drop > 0:
    df = df[~invalid_mask]
    print(f"Dropped {rows_to_drop} row(s) with illogical values")
else:
    print("No rows with illogical values found to drop")

# Final row count
print(f"Final number of rows: {len(df)}")
````

````python
# Write a function to calculate and analyse the target variable likelihood for categorical features


# First, create functions to separate plotting from analysis
def create_target_likelihood_plots(df, categorical_cols, target_column, target_value='Y', min_count=5):
    """
    Create plots for all categorical features showing their relationship with the target variable.
    Displays 5 plots per row.
    """
    # Calculate layout dimensions
    n_plots = len(categorical_cols)
    n_cols = 5  # 5 plots per row
    n_rows = (n_plots + n_cols - 1) // n_cols  # Ceiling division
    
    # Create a figure with subplots
    fig, axes = plt.subplots(n_rows, n_cols, figsize=(20, 4*n_rows))
    
    # Flatten axes array for easy indexing
    axes = axes.flatten() if n_rows > 1 else [axes] if n_plots == 1 else axes
    
    # Create each plot
    for i, col in enumerate(categorical_cols):
        if i < len(axes):
            ax = axes[i]
            
            # Create a crosstab of category vs target
            cross_tab = pd.crosstab(df[col], df[target_column])
            
            # Calculate the likelihood of target event for each category
            likelihood_df = pd.DataFrame({
                'Count': cross_tab.sum(axis=1),
                'Fraud_Count': cross_tab[target_value],
                'Fraud_Rate': cross_tab[target_value] / cross_tab.sum(axis=1)
            })
            
            # Sort by fraud rate to identify high-risk categories
            likelihood_df = likelihood_df.sort_values('Fraud_Rate', ascending=False)
            
            # Filter out categories with low counts for reliability
            reliable_categories = likelihood_df[likelihood_df['Count'] >= min_count]
            
            # Calculate overall fraud rate for reference
            overall_fraud_rate = df[target_column].value_counts(normalize=True)[target_value]
            
            # Filter to categories with sufficient counts for plotting
            if len(reliable_categories) > 15:
                # If too many categories, show only top and bottom ones
                top_n = 8
                plot_data = pd.concat([
                    reliable_categories.head(top_n),
                    reliable_categories.tail(top_n)
                ])
            else:
                plot_data = reliable_categories
            
            # Sort by fraud rate for the plot
            plot_data = plot_data.sort_values('Fraud_Rate')
            
            # Bar plot of fraud rates
            bars = ax.barh(y=plot_data.index, width=plot_data['Fraud_Rate'], color='skyblue')
            
            # Add a reference line for overall fraud rate
            ax.axvline(x=overall_fraud_rate, color='red', linestyle='--', 
                      label=f'Overall Rate: {overall_fraud_rate:.4f}')
            
            # Add value labels (with smaller font)
            for j, (_, row) in enumerate(plot_data.iterrows()):
                # Display fraud rate next to each bar
                ax.text(row['Fraud_Rate'] + 0.01, j, 
                        f"{row['Fraud_Rate']:.3f}", 
                        va='center', fontsize=8)
            
            # Formatting
            ax.set_xlabel('Fraud Rate', fontsize=9)
            ax.set_title(f'{col}', fontsize=10)
            
            # Only include legend on the first plot
            if i == 0:
                ax.legend(fontsize=8)
    
    # Hide any unused subplots
    for i in range(len(categorical_cols), len(axes)):
        if i < len(axes):
            axes[i].set_visible(False)
    
    plt.tight_layout()
    plt.show()

def analyze_categorical_target_likelihood(df, categorical_cols, target_column, target_value='Y', min_count=5):
    """
    Analyze the relationship between categorical features and target variable.
    Returns a summary of statistical findings.
    """
    results = []
    
    for col in categorical_cols:
        print(f"\n{'='*80}")
        print(f"Target Likelihood Analysis for: {col}")
        print(f"{'='*80}")
        
        # Create a crosstab of category vs target
        cross_tab = pd.crosstab(df[col], df[target_column])
        
        # Calculate the likelihood of target event for each category
        likelihood_df = pd.DataFrame({
            'Count': cross_tab.sum(axis=1),
            'Fraud_Count': cross_tab[target_value],
            'Fraud_Rate': cross_tab[target_value] / cross_tab.sum(axis=1)
        })
        
        # Sort by fraud rate to identify high-risk categories
        likelihood_df = likelihood_df.sort_values('Fraud_Rate', ascending=False)
        
        # Calculate overall fraud rate for reference
        overall_fraud_rate = df[target_column].value_counts(normalize=True)[target_value]
        
        # Display the likelihood table
        print(f"\nLikelihood Table for {col}:")
        print(f"Overall fraud rate: {overall_fraud_rate:.4f} ({overall_fraud_rate*100:.2f}%)")
        print("\nCategorical levels sorted by fraud rate:")
        display(likelihood_df)
        
        # Statistical test (Chi-square) to measure association
        chi2, p, dof, expected = stats.chi2_contingency(cross_tab)
        cramer_v = np.sqrt(chi2 / (df.shape[0] * (min(cross_tab.shape) - 1)))
        
        print(f"\nStatistical Association:")
        print(f"Chi-square statistic: {chi2:.4f}")
        print(f"p-value: {p:.4f}")
        print(f"Cramer's V: {cramer_v:.4f} (strength of association)")
        
        if p < 0.05:
            print(f"Significant association detected (p < 0.05)")
            if cramer_v < 0.1:
                print("But the association is weak (Cramer's V < 0.1)")
            elif cramer_v < 0.3:
                print("The association is moderate (0.1 <= Cramer's V < 0.3)")
            else:
                print("The association is strong (Cramer's V >= 0.3)")
        else:
            print(f"No significant association detected (p >= 0.05)")
        
        # Add results to list
        results.append({
            'column': col,
            'p_value': p,
            'cramers_v': cramer_v,
            'significant': p < 0.05,
            'strength': 'Strong' if cramer_v >= 0.3 else 'Moderate' if cramer_v >= 0.1 else 'Weak'
        })
    
    return results

# Get categorical columns
categorical_cols = X_train.select_dtypes(include=['category', 'object']).columns.tolist()

# Create combined dataframe for analysis
analysis_df = pd.concat([X_train, y_train], axis=1)

# First: create and display the plots (5 per row)
create_target_likelihood_plots(analysis_df, categorical_cols, 'fraud_reported')

# Second: perform detailed analysis and print statistics
results = analyze_categorical_target_likelihood(analysis_df, categorical_cols, 'fraud_reported')

# Summarize results
print("\n\nSUMMARY OF CATEGORICAL FEATURES' RELATIONSHIP WITH FRAUD:")
print("=" * 80)
# Sort by strength of association
results_df = pd.DataFrame(results).sort_values(['significant', 'cramers_v'], ascending=[False, False])
display(results_df)

# Identify categorical features with little predictive power
weak_features = results_df[(results_df['significant'] == False) | 
                         ((results_df['significant'] == True) & (results_df['strength'] == 'Weak'))]

if len(weak_features) > 0:
    print("\nCategorical features with limited predictive power:")
    for col in weak_features['column']:
        print(f"- {col}")
else:
    print("\nAll categorical features show significant association with the target variable.")
````

````python
# Write a function to calculate and analyse the target variable likelihood for categorical features

# Function to create plots for categorical target likelihood (5 plots per row)
def plot_categorical_target_likelihood(df, categorical_cols, target_column, target_value='Y', min_count=5):
    """
    Create visualizations of categorical features vs target variable with 5 plots per row.
    """
    # Calculate layout dimensions
    n_plots = len(categorical_cols)
    n_cols_per_row = 5  # 5 plots per row
    n_rows = (n_plots + n_cols_per_row - 1) // n_cols_per_row  # Ceiling division
    
    # Create a figure with subplots
    fig, axes = plt.subplots(n_rows, n_cols_per_row, figsize=(20, 3.5*n_rows))
    
    # Flatten axes array for easy indexing
    axes = axes.flatten() if n_rows > 1 else [axes] if n_plots == 1 else axes
    
    # Loop through each categorical column
    for i, col in enumerate(categorical_cols):
        if i < len(axes):
            ax = axes[i]
            
            # Create a crosstab of category vs target
            cross_tab = pd.crosstab(df[col], df[target_column])
            
            # Calculate the likelihood of target event for each category
            likelihood_df = pd.DataFrame({
                'Count': cross_tab.sum(axis=1),
                'Target_Count': cross_tab[target_value],
                'Target_Rate': cross_tab[target_value] / cross_tab.sum(axis=1)
            })
            
            # Sort by target rate
            likelihood_df = likelihood_df.sort_values('Target_Rate', ascending=False)
            
            # Filter out categories with low counts for reliability
            reliable_categories = likelihood_df[likelihood_df['Count'] >= min_count]
            
            # Calculate overall target rate for reference
            overall_rate = df[target_column].value_counts(normalize=True)[target_value]
            
            # Filter to categories with sufficient counts for plotting
            if len(reliable_categories) > 10:
                # If too many categories, show only top and bottom ones
                top_n = 5
                plot_data = pd.concat([
                    reliable_categories.head(top_n),
                    reliable_categories.tail(top_n)
                ])
            else:
                plot_data = reliable_categories
            
            # Sort by target rate for the plot
            plot_data = plot_data.sort_values('Target_Rate')
            
            # Bar plot of target rates
            bars = ax.barh(y=plot_data.index, width=plot_data['Target_Rate'], color='skyblue')
            
            # Add a reference line for overall target rate
            ax.axvline(x=overall_rate, color='red', linestyle='--', 
                      label=f'Overall: {overall_rate:.3f}')
            
            # Add value labels (with smaller font)
            for j, (_, row) in enumerate(plot_data.iterrows()):
                # Calculate position - avoid placing text outside the plot area
                text_x = min(row['Target_Rate'] + 0.01, 0.95)
                ax.text(text_x, j, f"{row['Target_Rate']:.3f}", va='center', fontsize=7)
            
            # Formatting
            ax.set_xlabel('Target Rate', fontsize=8)
            ax.set_title(f'{col}', fontsize=9)
            ax.tick_params(axis='both', which='major', labelsize=7)
            
            # Only include legend on the first plot
            if i == 0:
                ax.legend(fontsize=7)
    
    # Hide any unused subplots
    for i in range(len(categorical_cols), len(axes)):
        if i < len(axes):
            axes[i].set_visible(False)
    
    plt.tight_layout()
    plt.show()

# Make sure the original analyze_categorical_target_likelihood function is correctly defined
def analyze_categorical_target_likelihood(df, cat_column, target_column, target_value='Y', min_count=5):
    """
    Analyze the relationship between a categorical feature and target variable.
    
    Parameters:
    -----------
    df : pandas DataFrame
        The dataset containing features and target
    cat_column : str
        The name of the categorical column to analyze
    target_column : str
        The name of the target column
    target_value : str
        The value in target column representing the positive class
    min_count : int
        Minimum count of observations for a category to be included in analysis
    """
    print(f"\n{'='*80}")
    print(f"Target Likelihood Analysis for: {cat_column}")
    print(f"{'='*80}")
    
    # Create a crosstab of category vs target
    cross_tab = pd.crosstab(df[cat_column], df[target_column])
    
    # Calculate the likelihood of target event for each category
    likelihood_df = pd.DataFrame({
        'Count': cross_tab.sum(axis=1),
        'Fraud_Count': cross_tab[target_value],
        'Fraud_Rate': cross_tab[target_value] / cross_tab.sum(axis=1)
    })
    
    # Sort by fraud rate to identify high-risk categories
    likelihood_df = likelihood_df.sort_values('Fraud_Rate', ascending=False)
    
    # Filter out categories with low counts for reliability
    reliable_categories = likelihood_df[likelihood_df['Count'] >= min_count]
    
    # Calculate overall fraud rate for reference
    overall_fraud_rate = df[target_column].value_counts(normalize=True)[target_value]
    
    # Display the likelihood table
    print(f"\nLikelihood Table for {cat_column}:")
    print(f"Overall fraud rate: {overall_fraud_rate:.4f} ({overall_fraud_rate*100:.2f}%)")
    print("\nCategorical levels sorted by fraud rate:")
    display(likelihood_df)
    
    # Statistical test (Chi-square) to measure association
    chi2, p, dof, expected = stats.chi2_contingency(cross_tab)
    cramer_v = np.sqrt(chi2 / (df.shape[0] * (min(cross_tab.shape) - 1)))
    
    print(f"\nStatistical Association:")
    print(f"Chi-square statistic: {chi2:.4f}")
    print(f"p-value: {p:.4f}")
    print(f"Cramer's V: {cramer_v:.4f} (strength of association)")
    
    if p < 0.05:
        print(f"Significant association detected (p < 0.05)")
        if cramer_v < 0.1:
            print("But the association is weak (Cramer's V < 0.1)")
        elif cramer_v < 0.3:
            print("The association is moderate (0.1 <= Cramer's V < 0.3)")
        else:
            print("The association is strong (Cramer's V >= 0.3)")
    else:
        print(f"No significant association detected (p >= 0.05)")
    
    # Return insights
    return {
        'column': cat_column,
        'p_value': p,
        'cramers_v': cramer_v,
        'significant': p < 0.05,
        'strength': 'Strong' if cramer_v >= 0.3 else 'Moderate' if cramer_v >= 0.1 else 'Weak'
    }

# Get categorical columns from test data
categorical_cols_test = X_test.select_dtypes(include=['category', 'object']).columns.tolist()

# First, create the visualization with 5 plots per row
test_data = pd.concat([X_test, y_test], axis=1)
plot_categorical_target_likelihood(test_data, categorical_cols_test, 'fraud_reported')

# Now run the analysis and show statistics for each feature
results_test = []
for col in categorical_cols_test:
    result = analyze_categorical_target_likelihood(
        pd.concat([X_test[col], y_test], axis=1),
        col,
        'fraud_reported'
    )
    results_test.append(result)

# Summarize results
print("\n\nSUMMARY OF CATEGORICAL FEATURES' RELATIONSHIP WITH FRAUD (TEST DATA):")
print("=" * 80)
# Sort by strength of association
results_test_df = pd.DataFrame(results_test).sort_values(['significant', 'cramers_v'], ascending=[False, False])
display(results_test_df)
````

### **6.1 Perform resampling** <font color="red">[3 Marks]</font>

Handle class imbalance in the training data by applying resampling technique.

Use the **RandomOverSampler** technique to balance the data and handle class imbalance. This method increases the number of samples in the minority class by randomly duplicating them, creating synthetic data points with similar characteristics. This helps prevent the model from being biased toward the majority class and improves its ability to predict the minority class more accurately.

**Note:** You can try other resampling techniques to handle class imbalance

````python
# Section 6.1 (Modified)

# Import SMOTE
from imblearn.over_sampling import SMOTE

# --- START: Move Date Feature Creation Here ---
# Create date features on the ORIGINAL X_train and X_test first
print("Creating date features on original train/test sets...")
X_train['incident_month'] = X_train['incident_date'].dt.month
X_train['incident_day_of_week'] = X_train['incident_date'].dt.dayofweek
X_test['incident_month'] = X_test['incident_date'].dt.month
X_test['incident_day_of_week'] = X_test['incident_date'].dt.dayofweek
print("Date features created.")
# --- END: Move Date Feature Creation Here ---

# Initialize SMOTE
sampler = SMOTE(random_state=42, k_neighbors=5)

# Perform resampling on training data
print("\nPreparing data for resampling...")
print("Class distribution before resampling:")
print(y_train.value_counts())
print(f"Class imbalance ratio: {y_train.value_counts().iloc[0] / y_train.value_counts().iloc[1]:.2f}:1")

# Identify categorical columns (excluding the newly created numerical date features)
categorical_cols = X_train.select_dtypes(include=['category']).columns.tolist()

# Create a copy of X_train to work with (it now includes the new date features)
X_train_for_resampling = X_train.copy()

# Drop the original datetime column BEFORE resampling
if 'incident_date' in X_train_for_resampling.columns:
    X_train_for_resampling = X_train_for_resampling.drop('incident_date', axis=1)
    print("Dropped 'incident_date' column before resampling.")

# Convert categorical columns to integers
for col in categorical_cols:
    if col in X_train_for_resampling.columns: # Check if column still exists
        X_train_for_resampling[col] = X_train_for_resampling[col].cat.codes

# Apply the resampling
print("Applying SMOTE...")
X_resampled_np, y_resampled = sampler.fit_resample(X_train_for_resampling, y_train) # Note: fit_resample might return NumPy array
print("SMOTE finished.")

# Reconstruct X_resampled as a DataFrame with correct column names
X_resampled = pd.DataFrame(X_resampled_np, columns=X_train_for_resampling.columns)

# Convert the integer categorical columns back to their original category type
print("Converting categorical codes back to categories...")
for col in categorical_cols:
    if col in X_resampled.columns: # Check if column exists
        original_categories = X_train[col].cat.categories
        # Ensure codes are integers
        codes = X_resampled[col].round().astype(int)
        # Ensure codes are within the valid range for the categories
        valid_codes_mask = (codes >= 0) & (codes < len(original_categories))
        if not valid_codes_mask.all():
             print(f"Warning: Found invalid codes in column '{col}'. Clamping to valid range.")
             codes = np.clip(codes, 0, len(original_categories) - 1)

        X_resampled[col] = pd.Categorical.from_codes(
            codes,
            categories=original_categories
        )
print("Categorical conversion finished.")

# Check the class distribution after resampling
print("\nClass distribution after resampling:")
print(y_resampled.value_counts())
print(f"Class imbalance ratio: {y_resampled.value_counts().iloc[0] / y_resampled.value_counts().iloc[1]:.2f}:1")

# Check the shape of data before and after resampling
print(f"\nShape before resampling (X_train, incl. new date features): {X_train.shape}")
print(f"Shape after resampling (X_resampled): {X_resampled.shape}")

# --- Plotting code remains the same ---
# Create a bar plot to visualize the change in class distribution
plt.figure(figsize=(12, 6))
# Before resampling
plt.subplot(1, 2, 1)
sns.countplot(x=y_train, palette='viridis')
plt.title('Class Distribution Before Resampling')
plt.xlabel('Fraud Reported')
plt.ylabel('Count')
# After resampling
plt.subplot(1, 2, 2)
sns.countplot(x=y_resampled, palette='viridis')
plt.title('Class Distribution After Resampling')
plt.xlabel('Fraud Reported')
plt.ylabel('Count')
plt.tight_layout()
plt.show()
````

### **6.4 Combine values in Categorical Columns** <font color="red">[6 Marks]</font>
During the EDA process, categorical columns with multiple unique values may be identified. To enhance model performance, it is essential to refine these categorical features by grouping values that have low frequency or provide limited predictive information.

Combine categories that occur infrequently or exhibit similar behavior to reduce sparsity and improve model generalisation.

````python
# Combine categories that have low frequency or provide limited predictive information

# Function to combine infrequent categories
def combine_rare_categories(df, column, threshold=0.05, new_category='Other'):
    """
    Combine categories with frequency below threshold into a new 'Other' category
    
    Parameters:
    -----------
    df : pandas DataFrame
        The dataframe containing the column
    column : str
        Name of the categorical column
    threshold : float
        Frequency threshold below which categories will be combined (proportion)
    new_category : str
        Name for the new category that combines rare values
        
    Returns:
    --------
    pandas Series
        The column with combined categories
    """
    # Calculate frequency distribution
    value_counts = df[column].value_counts(normalize=True)
    
    # Identify rare categories (below threshold)
    rare_categories = value_counts[value_counts < threshold].index.tolist()
    
    if rare_categories:
        print(f"Column '{column}': Combining {len(rare_categories)} rare categories into '{new_category}'")
        print(f"  Rare categories: {rare_categories}")
        
        # Create a new series with combined categories
        combined_column = df[column].copy()
        combined_column = combined_column.cat.add_categories([new_category]) if new_category not in combined_column.cat.categories else combined_column
        combined_column = combined_column.replace(rare_categories, new_category)
        
        # Remove unused categories
        combined_column = combined_column.cat.remove_unused_categories()
        
        return combined_column
    else:
        print(f"Column '{column}': No categories below threshold {threshold*100}%, no combination needed")
        return df[column]

# Identify categorical columns
categorical_columns = X_resampled.select_dtypes(include=['category']).columns.tolist()
print(f"Processing {len(categorical_columns)} categorical columns")

# Process each categorical column in training data
for col in categorical_columns:
    # Skip columns with very few categories
    if X_resampled[col].nunique() <= 3:
        print(f"Column '{col}': Only {X_resampled[col].nunique()} categories, skipping")
        continue
        
    # Combine rare categories
    X_resampled[col] = combine_rare_categories(X_resampled, col, threshold=0.05)
    
    # Ensure test data has the same categories, combining as needed
    # First, ensure all categories from training exist in test
    test_categories = set(X_test[col].cat.categories)
    train_categories = set(X_resampled[col].cat.categories)
    
    missing_in_test = train_categories - test_categories
    if missing_in_test:
        X_test[col] = X_test[col].cat.add_categories(list(missing_in_test))
    
    # Then combine rare categories in test data using the same approach
    X_test[col] = combine_rare_categories(X_test, col, threshold=0.05)

# Display the results
print("\nAfter combining rare categories:")
for col in categorical_columns:
    if X_resampled[col].nunique() > 3:  # Only show columns with more than 3 categories
        print(f"\nColumn '{col}' category distribution:")
        display(X_resampled[col].value_counts(normalize=True).head(10))

````

#### **7.3.1** Plot ROC Curve  to visualise the trade-off between true positive rate and false positive rate across different classification thresholds <font color="red">[2 Marks]</font>

````python
# Import libraries or function to plot the ROC curve

from sklearn.metrics import roc_curve, roc_auc_score, make_scorer
import matplotlib.pyplot as plt
import numpy as np


# Define ROC function

def plot_roc_curve(y_true, y_prob):
    # Calculate false positive rate, true positive rate, and thresholds
    fpr, tpr, thresholds = roc_curve(y_true, y_prob)
    
    # Calculate AUC (Area Under Curve)
    auc = roc_auc_score(y_true, y_prob)
    
    # Plot the ROC curve
    plt.figure(figsize=(10, 8))
    plt.plot(fpr, tpr, color='blue', lw=2, label=f'ROC curve (AUC = {auc:.4f})')
    plt.plot([0, 1], [0, 1], color='gray', linestyle='--', label='Random Guess')
    plt.xlim([0.0, 1.0])
    plt.ylim([0.0, 1.05])
    plt.xlabel('False Positive Rate (1 - Specificity)')
    plt.ylabel('True Positive Rate (Sensitivity)')
    plt.title('Receiver Operating Characteristic (ROC) Curve')
    plt.legend(loc='lower right')
    plt.grid(True, alpha=0.3)
    
    # Return important values
    return fpr, tpr, thresholds, auc
````

````python
# Use grid search to find the best hyperparamter values

# Define the parameter grid to search
param_grid = {
    'n_estimators': [50, 100],                 # Number of trees
    'max_depth': [3,5,7],                # Depth of trees
    'min_samples_split': [10,20,30],              # Min samples to split node
    'min_samples_leaf': [5, 10, 15],                 # Min samples at leaf
    'max_features': ['sqrt', 'log2', None]         # Number of features for best split
}

# Create the random forest model for grid search
rf_for_grid = RandomForestClassifier(
    random_state=42,
    class_weight='balanced',
    n_jobs=-1
)


# # Outer CV for evaluation, inner CV for tuning
# outer_cv = KFold(n_splits=5, shuffle=True, random_state=42)
# inner_cv = KFold(n_splits=3, shuffle=True, random_state=42)


# Initialize GridSearchCV
# Use 3-fold cross-validation to save time, but still get reliable results
grid_search = GridSearchCV(
    estimator=rf_for_grid,
    param_grid=param_grid,
    cv=5,                        # Number of cross-validation folds
    scoring='f1',               # Use F1 score which balances precision and recall
    verbose=1,                   # Progress updates
    n_jobs=-1                    # Use all CPU cores
)

print("Starting Grid Search for Random Forest hyperparameters...")
print("This may take some time...")

# Fit the grid search
grid_search.fit(X_train_selected_rf, y_train_encoded)

print("Grid Search completed!")


###################################################################################

# Best Hyperparameters

# Display the best parameters found by grid search
print("Best Hyperparameters:")
for param, value in grid_search.best_params_.items():
    print(f"{param}: {value}")

# Display the best score
print(f"\nBest F1 Score from Grid Search: {grid_search.best_score_:.4f}")

# Compare with our previous model's cross-validation score
print(f"Previous model cross-validation F1 Score: {cv_f1.mean():.4f}")
print(f"Improvement: {grid_search.best_score_ - cv_f1.mean():.4f}")

# Display the full results table
results = pd.DataFrame(grid_search.cv_results_)

# Sort by rank
top_results = results.sort_values('rank_test_score').head(10)
print("\nTop 10 Parameter Combinations:")
display(top_results[['params', 'mean_test_score', 'std_test_score', 'rank_test_score']])
````
